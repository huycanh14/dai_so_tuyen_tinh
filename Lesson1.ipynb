## Một số cách khởi tạo khác

```python
# Mảng giá trị bằng 0
A1 = np.zeros(5) # Mảng 1 chiều, giá trị khởi tạo là 0 hết
print(A1)

print("=" * 20)

A2 = np.zeros((2, 3)) # Mảng (2x4), giá trị khởi tạo là 0
print(A2)

print("=" * 20)

# Taọ ra mảng giá trị khởi tạo bằng 1
A3 = np.ones((2, 3))
print(A3)
print("=" * 20)

A4 = 1.21 * A3
print(A4)
print("=" * 20)

A5 = np.full((2, 3), 5.) # Mảng (2x4), giá trị khởi tạo là 5
print(A5)
print("=" * 20)
```

```python
# Arange
start = 1
stop = 10
step = 2
A6 = np.arange(start, stop, step) # mảng 1 chiều: dãy số bắt bang 1 den 10, step = 2
print(A6)
print("=" * 20)

number = 20
A7 = np.linspace(start, stop, number) # tu step -> stop chia deu de co number so
print(A7)
print("=" * 20)
```

```python
# sinh ngau nhien
A8 = np.random.random((2, 3))  # sinh nhau nhien ma tran co kich thuoc 2x3
print(A8)
print("=" * 20)

# sinh nhau nhien ma tran co kich thuoc 2x3, nhung dai gia tri là so nguyen (gia tri từ 0 -> 45)
A9 = np.random.randint(0, 45, (2, 3))
print(A9)
print("=" * 20)

# sinh nhau nhien ma tran co kich thuoc 2x3: Gia tri trung binh, phuong sai, kich thuoc ma tran
A10 = np.random.normal(0, 9, (2, 3))
print(A10)
print("=" * 20)

```

## Truy cập phần tử của mảng

```python
# dùng 2 chỉ số: hàng và cột
arr_1 = np.random.random(6)
print(arr_1)
print(arr_1[3])
print("=" * 20)

arr_2 = np.random.random((2, 3))
print(arr_2)
print(arr_2[1, 2])
print("=" * 20)

# trích xuất mảng con
'''
x[start:stop:step]
'''
arr_3 = np.random.random(6)
print(arr_3)
print("Mảng con \n")
print(arr_3[3:1: -1])
print("=" * 20)

arr_4 = np.array([
    [-2, 1, 4, 5],
    [3, 2, -5, 2]
])
print(arr_4)
print("Mảng con \n")
print(arr_4[0, 3: 0: -1]) # hang dau tien, lay 3 phan tu, step = 1
print("=" * 20)
```

## Thay đổi kích thước mảng

```python
arr_st = np.arange(20)
print(arr_st)
print("Reshape to (4x5)")
print(arr_st.reshape((4, 5)))
print("=" * 20)
```

```python
A = np.random.randint(-10, 10, (10, 10))
print(A)
print("=" * 20)

A[:, 0] = A[0, 0]
print("Cac phan tu o cot dau tien thay doi")
print(A)
print("=" * 20)

A[:, -1] = A[0, 9]
print("Cac phan tu o cot cuoi cung thay doi")
print(A)
print("=" * 20)
```

```python
for i in range(1, 10):
    for j in range(1, 9):
        A[i, j] = (A[i - 1, j - 1] + A[i - 1, j] + A[i - 1, j + 1]) // 3

print(A)
print("=" * 20)
```

## Bài 2

![image info](./images/L1/b2.png)

```python
# 1. Tạo một mảng 1 chiều x gồm n = 10 số thực
n = 10
# Cách 1: sinh số thực đều trong khoảng [1, 10]
x = np.linspace(1, 10, n)
# (Hoặc cách 2: x = np.random.uniform(1.0, 10.0, n))
print("1. Mảng 1 chiều x (n = 10 số thực):")
print("x =", x)
print("x.shape =", x.shape)
print("=" * 40)

# 2. Chọn 3 hàm số tùy ý f1(x), f2(x), f3(x)
def f1(x):
    return 2 * x + 1    # Hàm bậc nhất

def f2(x):
    return x ** 2       # Hàm bậc hai

def f3(x):
    return np.sin(x)    # Hàm lượng giác

print("2. Ba hàm số đã chọn:")
print("f1(x) = 2x + 1")
print("f2(x) = x^2")
print("f3(x) = sin(x)")
print("=" * 40)

# 3. Tạo 3 mảng 1-chiều y1, y2, y3
y1 = f1(x)
y2 = f2(x)
y3 = f3(x)
print("3. Ba mảng 1 chiều y1, y2, y3:")
print("y1 =", y1)
print("y2 =", y2)
print("y3 =", y3)
print("=" * 40)

# 4. Ghép bốn mảng 1-chiều thành một mảng 2-chiều kích thước n x 4
# Cách ghép các vector cột: np.column_stack hoặc np.stack(..., axis=1)
result = np.column_stack((x, y1, y2, y3))
print("4. Mảng 2 chiều kích thước n x 4 (ghép từ x, y1, y2, y3):")
print(result)
print("Kích thước của mảng 2 chiều:", result.shape)
print("=" * 40)
```
