## Khởi tạo ma trận

```python
"""
np.empty(shape, dtype, order): trả về một ma trận mà các phần tử chứa được khởi tạo
        shape: khích thước ma trận
        dtype: kiểu dữ liệu phần tử
        order: optional ('C', 'F')
=> giá trị là các số có sẵn có trong ô nhớ được cấp phát

np.eye(n, M, k, dtype): trả về 1 ma trận mà các phần từ thuộc đường chéo bằng 1, các phần tử còn lại bằng 0
        n: số hàng của ma trận
        M: số cột của ma trận, giá trị mặc định = n
        k: chỉ số đường chéo
        dtype: kieu du lieu phan tu

np.random.rand(shape): tra ve ma tran ma cac gia tri duoc sinh ngau nhien tu [0, 1)
"""
```

```python
import numpy as np
```

```python
A = np.eye(4, 6, -1)

print(A)
print("=" * 40)
```

```python
"""
np.diag(v): 1 ma tran duong cheo, trong do v la 1 danh sach/mot kieu tuong duong
"""

B = np.random.randint(-5, 5, 3)
print(B)
print("=" * 40)
print(np.diag(B))
print("=" * 40)

print("\nVi du khac")
C = np.random.randint(-10, 10, (3, 3))
print(C)
print("=" * 40)
print(np.diag(C))
print("=" * 40)
```

```python
"""
np.tril(A, k): ma tran tam giac duoi
np.triu(A, k): ma tran tam giac tren
    A: mang 2 chieu/tuong duong
    k: (optional) chi so duong cheo, mac dinh k = 0
"""

D = np.random.randint(-10, 10, (4, 4))
print(D)
print("=" * 40)
print("k=0")
print(np.tril(D, k=0))
print("=" * 40)
print("k=1")
print(np.tril(D, k=1))
print("=" * 40)
print("k=-1")
print(np.tril(D, k=-1))
print("=" * 40)
```

```python
# Tao ma tran gom 3 duong cheo tuong ung voi chi so tuong ung 0, 1 va -1 cua 1 ma tran

E = np.random.randint(-10, 10, (5, 5))
print(E)
print("=" * 40)

print("np.triu(E, -1) =")
print(np.triu(E, -1))
print("=" * 40)

print("np.tril(E, 1) =")
print(np.tril(E, 1))
print("=" * 40)

print("Cong np.triu(E, -1) + np.tril(E, 1) =")
print(np.triu(E, -1) + np.tril(E, 1))
print("=" * 40)

F = np.triu(E, -1) + np.tril(E, 1) - E
print("Ket qua =")
print(F)
print("=" * 40)
```

# Nhan ma tran

```python
arr_A = np.array([
    [1, -2, 3],
    [4, 1, 5],
    # [1, 0, -2]
])

arr_B = np.random.randint(-10, 10, (3, 3))

print("arr_A = ", arr_A, sep='\n')
print("arr_B = ", arr_B, sep='\n')

print("=" * 40)
print("A x B =")
print(np.matmul(arr_A, arr_B))
print(np.dot(arr_A, arr_B))
print(arr_A@arr_B)
print("=" * 40)
```

## Luy thua

```python
arr_C = np.random.randint(-10, 10, (3, 3))
print("arr_C = ", arr_C, sep='\n')
print("=" * 40)

arr_D = np.linalg.matrix_power(arr_C, 2)
print("arr_D = ", arr_D, sep='\n')
```

## Ma tran chuyen vi

```python
arr_E = np.random.randint(-10, 10, (3, 3))
print("arr_E = ", arr_E, sep='\n')
print("=" * 40)

print("Chuyen vi\n")
print(np.transpose(arr_E))
print("=" * 40)
print(arr_E.T)
```

```python
# Vet cua ma tran: tổng các phần tử nằm trên đường chéo chính
print("Vet ma tran: trace(arr_E) = ", np.trace(arr_E))
print("=" * 40)
```

## Dinh thuc ma tran

```python
arr_F = np.random.randint(-10, 10, (3, 3))
print("arr_F = ", arr_F, sep='\n')
print("=" * 40)
print("det(arr_F) = ", np.linalg.det(arr_F))
print("=" * 40)
```

```python
# Hang ma tran
arr_rank = np.random.randint(-10, 10, (3, 3))
print("arr_rank = ", arr_rank, sep='\n')
print("=" * 40)
print("rank(arr_rank) = ", np.linalg.matrix_rank(arr_rank), sep='\n')
print("=" * 40)
```
